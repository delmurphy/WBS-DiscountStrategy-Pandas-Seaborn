# Cleaning the Orderlines dataframe

Read in data

```python
import pandas as pd
```

```python
path = '../data/data_raw/'
orderlines_orig = pd.read_csv(path + 'orderlines.csv')
```

```python
ol = orderlines_orig.copy()
```

### Check for duplicates

```python
ol.duplicated().sum()
```

```python
ol['id'].duplicated().sum()
```

### Eyeball the data

```python
ol.info()
```

```python
ol.isna().sum()
```

```python
ol.describe()
```

- unit_price and date need to be reformatted to numeric and datetime, respectively
- max product_quantity is 999 - seems strange (but this is maybe from an order that is cancelled or still in shopping basket - investigate in Data Quality step)

### Reformat unit_price to numeric

```python
#remove first '.' in unit_price
dot_mask = ol['unit_price'].str.count('\.')==2
ol.loc[dot_mask, 'unit_price'] = ol.loc[dot_mask, 'unit_price'].str.replace('.', '', 1)
ol['unit_price_num'] = pd.to_numeric(ol['unit_price'])
```

```python
ol.sample(10)
```

```python
ol.info()
```

```python
ol.isna().sum()
```

```python
ol['unit_price_num'].describe()
```

Min value (-119) and max value (159989) look strange
(possibly these orders are from cancelled/shopping cart orders - deal with them in Data Quality step)

### Reformat date as datetime

```python
ol['date_dt'] = pd.to_datetime(ol['date'])
ol.sample(10)
```

```python
ol.info()
```

looks good!

### Save to csv

```python
ol.to_csv('data_cleaned/orderlines_cleaned.csv')
```
